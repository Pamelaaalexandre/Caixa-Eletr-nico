```python
# Sistema de Caixa Eletrônico
def depositar(saldo):
  print(" ========= Bem Vindo ao GEN-CAIXA ========")
  print(f"Seu saldo atual é de {saldo:.2f}")
  valor_deposito = float(input("Informe o valor do depósito: "))
  saldo = (saldo + valor_deposito)
  return saldo
  # fazer o codigo
  # informar ao usuário o saldo atual
  # perguntar qual o valor quer depositar
  # fazer o algoritimo de deposito (saldo + deposito)
  # informar o novo saldo

def sacar(saldo):
  print(" ========= Bem Vindo ao GEN-CAIXA ========")
  print(f"Seu saldo atual é de {saldo:.2f}")
  valor_saque = float(input("Informe o valor do saque: "))
  if valor_saque > saldo:
    print("Saldo insulficiente")
    return saldo

  else:
    saldo = (saldo - valor_saque)
    return saldo
  # fazer o código
  # informar ao usuário o saldo atual
  #perguntar qual valor quer sacar
  # fazer o algoritmo do saque (saldo - saque)
  # informar o novo saldo
  #opcional - se o saldo for menor que o saque, não permitir
```

```python
print(" ========= Bem Vindo ao GEN-CAIXA ========")
saldo = float(input("Informe seu saldo: "))

print("Escolha uma das opções")
print("1 - Ver Saldo")
print("2 - Depositar")
print("3 - Sacar")

print("Qualquer outra opção - sair")

opcao_escolhida = int(input(""))

if opcao_escolhida == 1:
  print(f"Seu saldo atual é de {saldo:.2f}")

elif opcao_escolhida == 2:
  saldo = depositar(saldo)
  print(f"Seu saldo atual é de {saldo}")

elif opcao_escolhida == 3:
  saldo = sacar(saldo)
  print(f"Seu saldo atual é de {saldo}")

else:
  print("Obrigado")
```
